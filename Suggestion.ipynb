```python
import pandas as pd
rating_dataframe = pd.read_csv('u.data', sep='\t', names=['user_id','item_id','rating','timestamp'])
rating_dataframe.tail(5)
```

```python
df_movie_rating_mean= movie_rating.groupby('title')['rating'].describe()['mean']
df_movie_rating_mean
```

```python
df_movie_rating_count = movie_rating.groupby('title')['rating'].describe()['count']
df_movie_rating_count
```

```python
df_movie_rating_countandmean = pd.merge(df_movie_rating_mean,df_movie_rating_count, on='title')
df_movie_rating_countandmean
```

```python
df_movie_rating_countandmean2 = pd.concat([df_movie_rating_count,df_movie_rating_mean],axis=1)
df_movie_rating_countandmean2
```
